# **DATA CLEANING PROJECT**


The link to the dataset:
https://github.com/KeithGalli/Olympics-Dataset

This dataset contains biographical information about Olympic athletes, including their names, dates and places of birth, height, weight, and other personal details.


The dataset contains missing values, inconsistent formatting, and unstructured information that needs to be cleaned and standardized before analysis.

# **Tasks**

The main objective is to prepare the dataset for further analysis by performing the following data cleaning tasks:

1) Handle Missing Values — Identify and appropriately process missing or incomplete records.

2) Clean Height and Weight Data — Correct inconsistent measurements, separate combined values, and resolve incorrectly placed data.

3) Standardize Dates — Convert dates of birth into a consistent format.

4) Clean Birthplace Information — Standardize location names and extract geographical information into separate columns.

5) Fix Data Types — Convert columns into appropriate numeric, date, or text formats.

6) Standardize Text Values — Remove unnecessary characters, extra spaces, and formatting inconsistencies.

7) Validate the Cleaned Dataset — Check data consistency and ensure the dataset is ready for analysis.

## **P.S**:

This project was done for practice rather than presentation, so the cleaning steps may not be easy to follow.

My main goal here was to practice some data cleaning techniques, including regular expressions, string manipulation, handling missing values, data type conversion, date parsing, text standardization, splitting and extracting information from columns, and identifying and correcting inconsistent values.
