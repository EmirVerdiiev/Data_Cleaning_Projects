```python
date_text  =df['order_date'].astype('string').str.strip()
```

```python
ambiguous_mask = date_text.str.match(
    r"^(0?[1-9]|1[0-2])[-/](0?[1-9]|1[0-2])[-/]\d{4}$", na = False
)
```

```python
df.loc[
    ambiguous_mask,
    ["line_id", "order_id", "order_date"]
]
```

```python
amb_order_ids = df.loc[ambiguous_mask, "order_id"]
```
